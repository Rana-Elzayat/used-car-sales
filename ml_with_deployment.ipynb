# Machine Learning — Car Price Prediction

## Final ML Objective

After exploring the dataset, we initially considered **Car Sale Status** as a classification target.  
We tested Logistic Regression, Decision Tree, and Random Forest, but the classification results were weak.

We therefore changed the target to **`Price-$`**, which represents the initial price of the car when it was listed for sale. This turns the problem into a **Regression** task.

### Final objective
> Predict the initial car listing price (`Price-$`) using the available characteristics of the car.


## What happened during the ML process?

### 1. Initial Classification attempt
We first tried to predict **Car Sale Status (Sold / Un Sold)**.

The class distribution was imbalanced:
- Un Sold: 78.35%
- Sold: 21.65%

The tested classification models produced low F1 scores, so this target did not provide strong predictive performance with the available features.

### 2. Target reassessment
We investigated another possible target: **`Price-$`**.

The correlation analysis showed much stronger relationships with several vehicle attributes, especially:
- `Purchased Price-$`: 0.9205
- `Engine Power-HP`: 0.4939
- `Number of Doors`: -0.3579
- `Manufactured Year`: -0.2906

Because `Price-$` is the initial listing price, it is also a meaningful business target.

### 3. Avoiding data leakage
For the final regression model, we removed information that should not be used when predicting the initial listing price, including post-sale information such as:
- `Car Sale Status`
- `Sold Date`
- `Sold Price-$`
- `Margin-%`
- `Sales Commission-$`
- `Sales Rating`
- `Feedback`

We also excluded `Purchased Price-$` so that the model would not depend directly on the purchase price.

### 4. Final regression models
We compared:
1. Linear Regression
2. Decision Tree Regressor
3. Random Forest Regressor

The models were evaluated using **MAE, RMSE, and R²**.


```python
df_reg = df.copy()

df_reg["Purchased Date"] = pd.to_datetime(
    df_reg["Purchased Date"],
    errors="coerce"
)

df_reg["Purchased Year"] = df_reg["Purchased Date"].dt.year

df_reg["Car Age"] = (
    df_reg["Purchased Year"] - df_reg["Manufactured Year"]
)

df_reg.drop(columns=["Purchased Date"], inplace=True)
```

```python
target = "Price-$"

excluded_columns = [
    "ID",
    "Purchased Price-$",
    "Car Name",
    "Distributor Name",
    "Sales Agent Name",
    "Car Sale Status",
    "Sold Date",
    "Sold Price-$",
    "Margin-%",
    "Sales Commission-$",
    "Sales Rating",
    "Feedback"
]

X = df_reg.drop(columns=[target] + excluded_columns)
y = df_reg[target]

```

```python
numeric_X = X.select_dtypes(include=np.number)

price_corr = (
    numeric_X
    .corrwith(y)
    .sort_values(ascending=False)
)

price_corr
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

```

```python
categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

numerical_features = X.select_dtypes(
    include=[np.number]
).columns.tolist()

print("Numerical features:", numerical_features)
print("Categorical features:", categorical_features)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_features
        ),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ]
)
```

## Regression Models

- **Linear Regression**
- **Decision Tree Regressor**
- **Random Forest Regressor**


```python
models = {
    "Linear Regression": Pipeline([
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]),

    "Decision Tree": Pipeline([
        ("preprocessor", preprocessor),
        ("model", DecisionTreeRegressor(
            max_depth=10,
            min_samples_split=10,
            random_state=42
        ))
    ]),

    "Random Forest": Pipeline([
        ("preprocessor", preprocessor),
        ("model", RandomForestRegressor(
            n_estimators=200,
            max_depth=15,
            min_samples_split=5,
            random_state=42,
            n_jobs=-1
        ))
    ])
}
```

```python
# Train models

results = []

for name, model in models.items():

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)

    results.append({
        "Model": name,
        "MAE": mae,
        "RMSE": rmse,
        "R²": r2
    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by=["R²", "MAE"],
    ascending=[False, True]
).reset_index(drop=True)

results_df.round(3)
```

### Observed final results

**Linear Regression was selected as the best model because it achieved the lowest MAE and RMSE while also achieving an R² of 1.0.**

> Note: The exceptionally high performance is a characteristic of this dataset. The target contains only 50 unique price values across 10,000 records,  Duplicate rows and duplicate feature combinations were also checked during investigation and were not found.


```python
results_plot = results_df.set_index("Model")

results_plot[["MAE", "RMSE"]].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Regression Models - Error Comparison")
plt.ylabel("Error")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
results_plot["R²"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Regression Models - R² Comparison")
plt.ylabel("R²")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
final_model = results_df.iloc[0]["Model"]
best_model = models[final_model]

best_model.fit(X_train, y_train)
y_pred = best_model.predict(X_test)

plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred)

plt.xlabel("Actual Price ($)")
plt.ylabel("Predicted Price ($)")
plt.title(f"Actual vs Predicted Car Prices - {final_model}")

min_price = min(y_test.min(), y_pred.min())
max_price = max(y_test.max(), y_pred.max())

plt.plot(
    [min_price, max_price],
    [min_price, max_price],
    linestyle="--"
)

plt.tight_layout()
plt.show()
```

```python
# Residual Analysis
residuals = y_test - y_pred

plt.figure(figsize=(8, 6))

plt.scatter(y_pred, residuals)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Predicted Price ($)")
plt.ylabel("Residuals")
plt.title("Residual Analysis")

plt.tight_layout()
plt.show()
```

## Final ML Conclusion

The original classification task attempted to predict whether a car was **Sold or Un Sold**, but the tested classification models produced weak predictive performance.

After reassessing the target variable, **`Price-$`** was selected because it represents the initial listing price and showed stronger relationships with the vehicle characteristics.

Three regression models were compared:

- Linear Regression
- Decision Tree Regressor
- Random Forest Regressor

**Linear Regression was the final selected model**, achieving approximately zero MAE and RMSE and an R² of 1.0 on the test set.

The residual analysis showed that the prediction errors were extremely close to zero.

```python
from datetime import datetime
import gradio as gr
import joblib
import pandas as pd

model = joblib.load("car_price_model.pkl")

location_options = [
    "North Carolina",
    "Tennessee",
    "Detroit",
    "Chicago",
    "Denver",
    "Columbus",
    "Tucson",
    "Texas",
    "New York",
    "Utah",
    "California",
    "San Jose",
    "Philadelphia",
    "Florida",
    "Oklahoma",
    "Portland",
    "Madison",
]

manufacturer_options = [
    "Maruti",
    "Mahindra",
    "Toyota",
    "Hyundai",
    "Tata",
    "Renault",
    "Kia",
    "Prazo",
]

car_type_options = ["SUV", "Hatchback", "Truck", "Sedan", "Convertible"]
color_options = ["Gray", "Blue", "Red", "Black", "White"]
gearbox_options = ["Automatic", "Manual"]
energy_options = ["Hybrid", "Diesel", "Petrol", "Electric"]


def predict(
    location,
    manufacturer,
    car_type,
    color,
    gearbox,
    energy,
    seats,
    doors,
    year,
    mileage,
    horsepower,
    purchased_date,
):

    purchased_date_dt = pd.to_datetime(purchased_date)
    today = pd.to_datetime(datetime.now().date())

    if purchased_date_dt > today:
        return """
        <div class='error-box'>
        Purchased date cannot be in the future.
        </div>
        """

    purchased_year = purchased_date_dt.year
    purchased_month = purchased_date_dt.month
    purchased_dayofweek = purchased_date_dt.dayofweek

    car_age = purchased_year - year

    if year > purchased_year:
        return """
        <div class='error-box'>
        Manufactured year cannot be greater than Purchased year.
        </div>
        """

    df = pd.DataFrame(
        [
            {
                "Location": location,
                "Manufacturer Name": manufacturer,
                "Car Type": car_type,
                "Color": color,
                "Gearbox": gearbox,
                "Energy": energy,
                "Number of Seats": seats,
                "Number of Doors": doors,
                "Manufactured Year": year,
                "Mileage-KM": mileage,
                "Engine Power-HP": horsepower,
                "Purchased Year": purchased_year,
                "Purchased Month": purchased_month,
                "Purchased DayOfWeek": purchased_dayofweek,
                "Car Age": car_age,
            }
        ]
    )

    prediction = model.predict(df)[0]

    return f"""
    <div class='result-card'>
        <h2>Estimated Vehicle Price</h2>
        <h1>${prediction:,.2f}</h1>

        <hr>

        <h3>Vehicle Summary</h3>

        <p><b>Manufacturer:</b> {manufacturer}</p>
        <p><b>Type:</b> {car_type}</p>
        <p><b>Color:</b> {color}</p>
        <p><b>Mileage:</b> {mileage:,} KM</p>
        <p><b>Age:</b> {car_age} Years</p>

        <hr>

        <h3>AI Insights</h3>

        <ul>
            <li>Prediction generated successfully.</li>
            <li>Model analyzed all 15 vehicle features.</li>
            <li>Estimated market listing price.</li>
        </ul>
    </div>
    """


css = """
body, .gradio-container {
    background-color: var(--background-fill-primary) !important;
}

.gradio-container {
    max-width: 1200px !important;
    margin: auto;
}

.hero {
    background: linear-gradient(90deg, #0f62fe, #0043ce);
    padding: 30px;
    border-radius: 18px;
    color: white !important;
    text-align: center;
    margin-bottom: 25px;
}

.cards {
    display: flex;
    gap: 20px;
    justify-content: center;
    margin-bottom: 25px;
    flex-wrap: wrap;
}

.card {
    background: var(--background-fill-secondary);
    border: 1px solid var(--border-color-primary);
    padding: 20px;
    border-radius: 15px;
    box-shadow: 0px 5px 15px rgba(0,0,0,0.1);
    width: 220px;
    text-align: center;
}

.card h2 {
    color: #ff6f00 !important;
    font-weight: bold;
}

.card p {
    color: var(--body-text-color) !important;
    font-weight: 500;
}

.result-card {
    background: var(--background-fill-secondary);
    border: 1px solid var(--border-color-primary);
    padding: 30px;
    border-radius: 20px;
    box-shadow: 0px 5px 20px rgba(0,0,0,0.15);
    color: var(--body-text-color) !important;
}

.result-card h1 {
    color: #0f62fe !important;
    font-size: 48px;
    font-weight: bold;
}

.result-card h2,
.result-card h3 {
    color: var(--body-text-color-subdued) !important;
    font-weight: bold;
}

.result-card p,
.result-card li,
.result-card span,
.result-card b,
.result-card div {
    color: var(--body-text-color) !important;
    font-size: 16px;
}

.result-card hr {
    border-color: var(--border-color-primary) !important;
    margin: 15px 0;
}

.error-box {
    background: #ffdede;
    padding: 20px;
    border-radius: 10px;
    color: #cc0000 !important;
    font-size: 20px;
    font-weight: bold;
}

footer {
    visibility: hidden;
}
"""


with gr.Blocks(css=css, title="AI Car Price Predictor") as demo:

    gr.HTML(
        """
    <div class='hero'>
        <h1>🚗 AI Car Price Predictor</h1>
        <h3>Machine Learning Based Used Car Valuation</h3>
        <p>
        Predict the estimated initial listing price using
        a trained regression model.
        </p>
    </div>
    """
    )

    gr.HTML(
        """
    <div class='cards'>
        <div class='card'>
        <h2>🤖 Model</h2>
        <p>Regression Pipeline</p>
        </div>

        <div class='card'>
        <h2>📊 Features</h2>
        <p>15 Inputs</p>
        </div>

        <div class='card'>
        <h2>⚡ Speed</h2>
        <p>Instant Prediction</p>
        </div>

        <div class='card'>
        <h2>🎯 Output</h2>
        <p>Estimated Price</p>
        </div>
    </div>
    """
    )

    with gr.Row():

        with gr.Column():

            gr.Markdown("## Vehicle Information")

            location = gr.Dropdown(location_options, label="Location")
            manufacturer = gr.Dropdown(
                manufacturer_options, label="Manufacturer"
            )
            car_type = gr.Dropdown(car_type_options, label="Car Type")
            color = gr.Dropdown(color_options, label="Color", value="Blue")
            gearbox = gr.Dropdown(gearbox_options, label="Gearbox")
            energy = gr.Dropdown(energy_options, label="Fuel Type")

        with gr.Column():

            gr.Markdown("## Technical Details")

            seats = gr.Slider(2, 8, value=5, step=1, label="Seats")
            doors = gr.Slider(2, 5, value=4, step=1, label="Doors")
            year = gr.Slider(
                2015, 2024, value=2020, step=1, label="Manufactured Year"
            )
            mileage = gr.Number(value=45000, label="Mileage (KM)")
            horsepower = gr.Number(value=120, label="Engine Power (HP)")
            purchased_date = gr.Textbox(
                value="2023-01-01", label="Purchased Date (YYYY-MM-DD)"
            )

    with gr.Row():

        predict_btn = gr.Button(
            "🚀 Predict Vehicle Price", variant="primary", size="lg"
        )

        clear_btn = gr.ClearButton()

    output = gr.HTML()

    predict_btn.click(
        predict,
        inputs=[
            location,
            manufacturer,
            car_type,
            color,
            gearbox,
            energy,
            seats,
            doors,
            year,
            mileage,
            horsepower,
            purchased_date,
        ],
        outputs=output,
    )

demo.launch(share=True)
```
